## Descrizione del progetto
L'azienda AutomaParts S.p.A. è un fornitore tier-1 per il settore automotive che produce componenti meccanici di precisione per sistemi sterzo e sospensioni. Il progetto ha l'obiettivo di sviluppare, usando Python e consegnando un notebook Google Colab, un modello di machine learning supervisionato per distinguere pezzi conformi da pezzi difettosi in linea di produzione. Il modello dovrà supportare decisioni operative per ridurre scarti, aumentare la qualità e migliorare i tempi di ispezione.

Il lavoro è pensato per essere applicabile a dati raccolti da celle di produzione reali (misure dimensionali, parametri di processo, segnali di sensori, informazioni di linea e operatore) e dovrà includere analisi del dataset, gestione dei dati mancanti, preparazione delle variabili, addestramento e valutazione di modelli (ad es. Importanza per l'azienda: riduzione dei costi di scarto e rilavorazione, aumento del First Pass Yield, diminuzione del rischio di non conformità verso OEM, e migliore efficienza della linea.

Valore aggiunto: tracciabilità dei motivi di scarto aggregata per macchina/lotto/operatore, supporto alle decisioni manutentive e potenziale integrazione con sistemi MES per azioni automatizzate.

## 3. Pulizia dei dati e gestione dei valori mancanti

**Obiettivo.** Definire come trattare valori mancanti e anomali, anche in assenza di missing nel dataset corrente.

Nel dataset attuale non presenta valori mancanti, tuttavia definiamo comunque una strategia di imputazione — mediana per le variabili numeriche, moda per le categoriche — da applicare se in produzione dovessero comparire letture mancanti. Nella fase di preprocessing, solo sul training set.

```python
# identifichiamo valori outlier

def count_outliers_iqr(series):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return int(((series < lower) | (series > upper)).sum())

pd.Series({c: count_outliers_iqr(df[c]) for c in num_features}, name="n_outliers_iqr").sort_values(ascending=False)
```

Il numero di valori anomali secondo la regola IQR è nullo o trascurabile per tutte le variabili. 
In un contesto di controllo qualità, un valore ai limiti di tolleranza è spesso l'osservazione più informativa per riconoscere un pezzo difettoso, non rumore da eliminare. Quindi decidiamo comunque di non rimuovere gli outliers.

## 4. Feature engineering e preparazione delle variabili

**Obiettivo.** Derivare feature aggiuntive da timestamp e lotto, definire encoding e verificare la ridondanza tra variabili numeriche.

```python
# creamo una copia del df in cui aggiungiamo colonne estrapolate dalla data: oraa, giorno, mese e droppiamo 
# le info non rilevanti come id, timestamp e codice del batch (da cui abbiamo già preso le info utili)

df_fe = df.copy()

# Time-derived features from the production timestamp
df_fe["production_hour"] = df_fe["production_timestamp"].dt.hour
df_fe["day_of_week"] = df_fe["production_timestamp"].dt.dayofweek
df_fe["production_month"] = df_fe["production_timestamp"].dt.month

# Production week of the batch, extracted from the batch code (e.g. "MB-2024W24-L02-575" -> 24)
df_fe["batch_week"] = df_fe["material_batch"].str.extract(r"W(\d+)").astype(int)

df_fe = df_fe.drop(columns=["part_id", "production_timestamp", "material_batch"])
df_fe.head()
```

`part_id` non ha potere predittivo generalizzabile. `production_timestamp` non è direttamente utilizzabile, ma ne estraiamo ora, giorno della settimana e mese. `material_batch` ha quasi tanti valori distinti quante righe (~2.997 su 3.000): usarlo come categoria diretta porterebbe a overfitting quasi totale, quindi ne estraiamo solo la settimana di produzione, un'informazione aggregata e riutilizzabile su lotti futuri.

```python
# separiamo le colonne con info categoriche da quelle numeriche

categorical_cols = ["line_id", "station_id", "operator_id"]
numeric_cols = [c for c in df_fe.columns if c not in categorical_cols + ["defect_label"]]

for c in categorical_cols:
    df_fe[c] = df_fe[c].astype(str)

print("Categorical features:", categorical_cols)
print("Numeric features:", numeric_cols)
```

`line_id`, `station_id` e `operator_id` sono identificativi nominali, senza un ordine naturale: li codifichiamo con One-Hot Encoding invece di un `LabelEncoder`, che imporrebbe una relazione d'ordine inesistente. `operator_id` resta un identificativo numerico anonimo, senza dati anagrafici, per il requisito di privacy della consegna.

```python
# e cerchiamo possibili correlazioni tra le colonne con dati numerici

corr_predictors = df_fe[numeric_cols].corr()
corr_pairs = corr_predictors.where(np.triu(np.ones(corr_predictors.shape), k=1).astype(bool)).stack()
corr_pairs.reindex(corr_pairs.abs().sort_values(ascending=False).index).head(5)
```

La correlazione massima tra due variabili predittive non supera 0.05: le misure di processo sono sostanzialmente indipendenti tra loro, quindi la multicollinearità non è una causa concreta del segnale debole osservato. Non è un problema ora, ma nella Regressione Logistica useremo comunque la regolarizzazione L2per proteggerci nel caso in futuro arrivino nuovi sensori correlati a quelli già presenti.

## 5. Suddivisione train/test e pipeline di preprocessing

**Obiettivo.** Separare training e test in modo rappresentativo e incapsulare la preparazione dei dati in una pipeline che eviti data leakage.

```python
# prepariamo la matrice con le features e l'array con i target, sia per il training che per il test

X = df_fe.drop(columns=["defect_label"])
y = df_fe["defect_label"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED, stratify=y
)

print("Train:", X_train.shape, "- Test:", X_test.shape)
print(f"Defect rate -> Train: {y_train.mean():.3f} - Test: {y_test.mean():.3f}")
```

Usiamo `stratify=y` perché i pezzi difettosi sono una minoranza (~22%): senza stratificazione un singolo split casuale potrebbe generare un test set con una proporzione di difetti non rappresentativa.

```python
# definiamo il preprocessor per riempire i dati numerici e categorici mancanti nelle varie righe,
# scala le variabili numeriche e fa OHE delle categoriche

preprocessor = ColumnTransformer(transformers=[
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_cols),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("ohe", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_cols)
])
```

Il preprocessore è incapsulato in una `Pipeline`: mediana, moda e parametri di scaling vengono appresi solo sul training set e poi applicati al test set, senza alcun leakage.

## 6. Ottimizzazione della Regressione Logistica: L1, L2 ed Elastic Net

**Obiettivo.** Esaminare le tre forme di regolarizzazione disponibili per la Regressione Logistica — L1 (Lasso), L2 (Ridge) ed Elastic Net (loro combinazione) — capire come agiscono sui coefficienti, e selezionare tramite cross-validation la configurazione da usare nel resto dello studio.

Usiamo `scoring="f1"` per identificare il miglior risultato della ricerca: con una classe minoritaria al 22% l'accuratezza sarebbe fuorviante, poiché un modello che predice sempre "conforme" otterrebbe già ~78% senza alcuna utilità pratica.

### 6.1 L1 (Lasso): selezione delle feature

La penalizzazione L1 spinge verso lo zero i coefficienti delle variabili meno utili, azzerandoli del tutto: agisce quindi come un selettore automatico di feature. Osserviamo quante ne azzera a una regolarizzazione fissa (`C=0.1`).

```python
# testiamo il comportamento della regressione con solo L1

feature_names = preprocessor.fit(X_train).get_feature_names_out()

l1_model = Pipeline([("prep", preprocessor),
                      ("clf", LogisticRegression(l1_ratio=1, C=0.1, solver="saga",
                                                  class_weight="balanced", max_iter=5000,
                                                  random_state=RANDOM_SEED))])
l1_model.fit(X_train, y_train)

l1_coef = l1_model.named_steps["clf"].coef_[0]
n_zero_l1 = int((l1_coef == 0).sum())

plt.figure(figsize=(8, 8))
plt.barh(feature_names, l1_coef, color="#C44E52")
plt.title("L1 (Lasso) coefficients - C=0.1")
plt.tight_layout()
plt.show()

print(f"Features zeroed by L1: {n_zero_l1} / {len(feature_names)}")
```

L1 azzera 37 dei 53 coefficienti delle features: meno di un terzo delle features riceve un peso non nullo. In un dataset con relazioni forti, questo indicherebbe che poche variabili concentrano l'informazione utile. Qui va letto con cautela: dato il segnale debole già osservato, l'azzeramento riflette soprattutto l'assenza di un contributo individuale netto della maggior parte delle features.

### 6.2 L2 (Ridge): riduzione uniforme

La penalizzazione L2 riduce l'ampiezza di tutti i coefficienti senza azzerarli: mantiene ogni feature nel modello, ma con un peso contenuto. È più stabile di L1 quando le variabili portano contributi piccoli e diffusi, e resiste meglio a variazioni minime nei dati di processo.

```python
# testiamo il comportamento della regressione con solo L2

l2_model = Pipeline([("prep", preprocessor),
                      ("clf", LogisticRegression(l1_ratio=0, C=0.1, solver="saga",
                                                  class_weight="balanced", max_iter=5000,
                                                  random_state=RANDOM_SEED))])
l2_model.fit(X_train, y_train)

l2_coef = l2_model.named_steps["clf"].coef_[0]
n_zero_l2 = int((l2_coef == 0).sum())

plt.figure(figsize=(8, 8))
plt.barh(feature_names, l2_coef, color="#4C72B0")
plt.title("L2 (Ridge) coefficients - C=0.1")
plt.tight_layout()
plt.show()

print(f"Features zeroed by L2: {n_zero_l2} / {len(feature_names)}")
```

Come atteso, L2 non azzera alcun coefficiente: tutte le 53 feature restano nel modello con pesi ridotti e distribuiti. Il contrasto con L1 (37 azzerati) illustra la differenza pratica tra le due penalizzazioni: L1 produce un modello sparso e interpretabile in termini di poche variabili, L2 un modello denso che sfrutta piccoli contributi da tutte.

### 6.3 Elastic Net: combinazione tramite `l1_ratio`

Elastic Net combina L1 e L2 in un'unica penalizzazione, regolata dal parametro `l1_ratio`: `l1_ratio=1` corrisponde a L1 puro, `l1_ratio=0` a L2 puro, i valori intermedi a una miscela delle due. Invece di scegliere a priori tra L1 e L2, lasciamo che la cross-validation individui il punto migliore lungo questo continuo, cercando contemporaneamente su `l1_ratio` e sulla forza di regolarizzazione `C`. È lo stesso parametro `l1_ratio` usato nelle due celle precedenti: qui non lo fissiamo, ma lo includiamo nella griglia di ricerca.

```python
# testiamo il comportamento della regressione ricercando la miglior combinazione tre L1 e L2 

en_pipeline = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(solver="saga", class_weight="balanced",
                                max_iter=5000, random_state=RANDOM_SEED))
])

param_grid = {
    "clf__C": [0.001, 0.01, 0.1, 0.5, 1, 10],
    "clf__l1_ratio": [0.0, 0.25, 0.5, 0.75, 1.0],
}

grid_log = GridSearchCV(en_pipeline, param_grid, cv=5, scoring="f1", n_jobs=-1)
grid_log.fit(X_train, y_train)

print("Best params:", grid_log.best_params_)
print(f"Best cross-validated F1: {grid_log.best_score_:.3f}")
```

```python
# plottiamo i risultati dei vari F1 calcolati con le differenti combinazioni di L1 e L2

cv_results = pd.DataFrame(grid_log.cv_results_)
f1_by_ratio = cv_results.groupby("param_clf__l1_ratio")["mean_test_score"].max()

plt.figure(figsize=(7, 4))
f1_by_ratio.plot(marker="o", color="#55A868")
plt.xlabel("l1_ratio  (0 = L2/Ridge,  1 = L1/Lasso)")
plt.ylabel("Best cross-validated F1")
plt.title("F1 vs l1_ratio - Elastic Net search")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

f1_by_ratio.round(3)
```

La ricerca seleziona `l1_ratio=0` con `C=0.01`: la cross-validation preferisce l'estremo L2 puro dell'intervallo Elastic Net. Il grafico mostra però che le differenze di F1 tra i vari valori di `l1_ratio` sono minime (tutte comprese in circa 0.34–0.35): coerente con quanto visto finora, nessuna forma di regolarizzazione offre un vantaggio marcato, perché il fattore limitante non è come si distribuisce la penalizzazione ma la debole informatività complessiva delle feature. La scelta di `C` basso conferma inoltre una regolarizzazione forte: il modello rinuncia ai dettagli fini del training set per non adattarsi al rumore.

Il modello finale della Regressione Logistica è quello individuato dalla ricerca Elastic Net (`grid_log.best_estimator_`), che nel nostro caso coincide con una L2 fortemente regolarizzata.

## 10. Decision threshold e trade-off precision/recall

**Obiettivo.** Tradurre il compromesso precision/recall in una decisione operativa, dando esplicitamente più peso al recall rispetto alla precision — coerente con l'idea che un falso negativo (difetto non intercettato) è più grave di un falso positivo (falso allarme).

`predict()` applica di default una soglia di 0.5 alla probabilità restituita da `predict_proba()`. Questa soglia è appropriata solo quando i due tipi di errore hanno la stessa importanza — non è il nostro caso: un falso negativo può causare un fermo linea o un richiamo lato cliente, un falso positivo comporta solo un controllo aggiuntivo. Per riflettere questa asimmetria usiamo l'F-beta score: un parametro `beta` che esprime quante volte il recall conta più della precision (`beta=1` equivale a F1, `beta=2` dà al recall il doppio del peso, e così via).

```python
# osserviamo i risultati al variare della probabilità di classificazione

y_proba_rf = fitted_models["Random Forest"].predict_proba(X_test)[:, 1]

for t in [0.50, 0.40, 0.30,0.20]:
    y_pred_t = (y_proba_rf >= t).astype(int)
    print(f"threshold={t}: accuracy={accuracy_score(y_test, y_pred_t):.3f} "
          f"precision={precision_score(y_test, y_pred_t, zero_division=0):.3f} "
          f"recall={recall_score(y_test, y_pred_t):.3f}")
```

Abbassare la soglia di pochi centesimi produce un effetto molto più brusco di quanto normalmente atteso. Le probabilità di Random Forest restano compresse in una fascia stretta attorno a 0.45–0.50, riflesso diretto del segnale debole già discusso. Una soglia scelta "a occhio" è quindi rischiosa come criterio operativo: la sistematizziamo collegandola esplicitamente al peso relativo che diamo al recall, tramite l'F-beta score.

```python
# valutiamo il risultato dando più importanza ai falsi negativi

BETA = 2  # recall has the double weight compared to precision: a false negative is way more damaging that a false positive

thresholds_scan = np.arange(0.05, 0.95, 0.01)
fbeta_scores = np.array([
    fbeta_score(y_test, (y_proba_rf >= t).astype(int), beta=BETA, zero_division=0)
    for t in thresholds_scan
])
best_threshold = thresholds_scan[fbeta_scores.argmax()]

plt.figure(figsize=(8, 5))
plt.plot(thresholds_scan, fbeta_scores, color="#4C72B0")
plt.axvline(0.5, color="gray", linestyle="--", label="Default threshold (0.5)")
plt.axvline(best_threshold, color="#C44E52", linestyle="--", label=f"F{BETA}-optimal threshold ({best_threshold:.2f})")
plt.xlabel("Decision threshold")
plt.ylabel(f"F{BETA} score")
plt.title(f"F{BETA} score vs threshold - Random Forest")
plt.legend()
plt.tight_layout()
plt.show()

print(f"F{BETA} at default threshold (0.5): {fbeta_scores[np.argmin(np.abs(thresholds_scan - 0.5))]:.3f}")
print(f"F{BETA} at optimal threshold ({best_threshold:.2f}): {fbeta_scores.max():.3f}")
```

Con `beta=2` (il recall pesa il doppio della precision), la soglia ottimale scende da 0.5 a 0.30, segnalando come "a rischio" quasi tutti i pezzi pur di non perderne nessuno. È un risultato scomodo ma informativo: **con questo potere predittivo, non è possibile automatizzare in modo affidabile lo scarto quando si dà priorità forte al recall** — la scelta "sicura" equivale quasi a controllare tutto manualmente.

```python
# valutiamo allora vari valori di beta

beta_values = [0.5, 1, 1.1, 1.2, 1.5, 2, 3, 5, 10]
sensitivity_rows = []
for beta in beta_values:
    fbeta_scan = np.array([
        fbeta_score(y_test, (y_proba_rf >= t).astype(int), beta=beta, zero_division=0)
        for t in thresholds_scan
    ])
    t_opt = thresholds_scan[fbeta_scan.argmax()]
    cm_opt = confusion_matrix(y_test, (y_proba_rf >= t_opt).astype(int))
    sensitivity_rows.append({"Beta (recall weight)": beta, "Optimal threshold": t_opt,
                              "Residual FN": cm_opt[1, 0], "Generated FP": cm_opt[0, 1]})

sensitivity_df = pd.DataFrame(sensitivity_rows)
sensitivity_df
```

La soglia ottimale resta stabile a 0.42 per `beta` fino a 1.2, poi scende a 0.30 e vi si ferma (**satura**) già a partire da `beta=1.5`: da quel punto i falsi negativi sul test set sono già a zero, quindi dare ancora più peso al recall (`beta=10` invece di `beta=1.5`) non sposta più la soglia — non c'è più nessun falso negativo da eliminare. La saturazione è essa stessa un'informazione operativa: oltre un certo punto, la leva efficace non è più quanto peso diamo al recall ma il potere predittivo del modello.

Una soluzione più realistica di una soglia unica è un sistema a tre zone, con i tagli derivati dalle stesse soglie F-beta ottimizzate sopra invece che da percentili arbitrari:

```python
# proponiamo una soluzione alternativa: fasce di valutazione invece di risposta binaria con Random forest (ROC-AUC più alto) 

# high_cut: F-beta 0.5-optimal threshold (precision weighted double) - read from the sensitivity table computed above
precision_optimal_threshold = sensitivity_df.loc[
    sensitivity_df["Beta (recall weight)"] == 0.5, "Optimal threshold"
].iloc[0]

low_cut = best_threshold                  # F-beta 2-optimal threshold (beta=2, cell above): below it, the model considers it safe even when prioritizing recall
high_cut = precision_optimal_threshold    # F-beta 0.5-optimal threshold: above it, the model is confident enough to reject

zone = np.where(y_proba_rf < low_cut, "Automatic acceptance",
        np.where(y_proba_rf >= high_cut, "Reject / full inspection", "100% inspection"))

zone_summary = pd.DataFrame({"zone": zone, "true_defect": y_test.values}).groupby("zone")["true_defect"].agg(
    n_pieces="count", defect_rate="mean")
zone_summary["defect_rate"] = (zone_summary["defect_rate"] * 100).round(1)
zone_summary.reindex(["Automatic acceptance", "100% inspection", "Reject / full inspection"])
```

Le due soglie non sono più percentuali scelte a mano, ma le stesse soglie F-beta ricavate dai dati nelle celle precedenti: `low_cut` è la soglia F-beta 2-ottimale (già usata per capire fin dove il modello può spingersi dando priorità al recall), `high_cut` è la soglia F-beta 0.5-ottimale (dove la precision pesa il doppio, quindi il modello è sicuro abbastanza da scartare senza generare troppi falsi allarmi). La dimensione di ciascuna zona e il relativo tasso di difetti dipendono ora direttamente da queste soglie — il modello non basta per una decisione binaria affidabile, ma resta utile per **allocare priorità di ispezione**: se ne raccomanda l'utilizzo in questa forma, non come sostituto del controllo qualità.

## 12. Verifiche di robustezza

**Obiettivo.** Verificare se le conclusioni raggiunte dipendono dai modelli scelti o dalla numerosità delle feature, oppure descrivano un limite reale del dataset. Due verifiche indipendenti: ridurre le feature a quelle più informative, e aumentare la capacità del modello con una rete neurale.

### 12.1 Feature selection

Riaddestriamo i tre modelli usando solo le 10 feature più informative individuate dal Random Forest, per verificare se le feature meno rilevanti stiano introducendo rumore che penalizza le prestazioni.

```python
#riaddestriamo i modelli solo sulle features più rilevanti per essi e valutiamo nuovamente i risultati

top_n = 10
selected_transformed = importances.head(top_n).index.tolist()
selected_original = sorted({original_feature_name(f, categorical_cols, numeric_cols) for f in selected_transformed})

selected_numeric = [c for c in selected_original if c in numeric_cols]
selected_categorical = [c for c in selected_original if c in categorical_cols]

print("Selected original columns:", selected_original)

preprocessor_fs = ColumnTransformer(transformers=[
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler())]), selected_numeric),
    ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")), ("ohe", OneHotEncoder(handle_unknown="ignore"))]), selected_categorical),
])

models_fs = {
    "Logistic Regression (FS)": Pipeline([("prep", preprocessor_fs),
                                           ("clf", LogisticRegression(**{k.replace("clf__", ""): v for k, v in grid_log.best_params_.items()},
                                                                       solver="saga", class_weight="balanced",
                                                                       random_state=RANDOM_SEED, max_iter=5000))]),
    "Decision Tree (FS)": Pipeline([("prep", preprocessor_fs),
                                     ("clf", DecisionTreeClassifier(max_depth=5, min_samples_leaf=20,
                                                                     class_weight="balanced", random_state=RANDOM_SEED))]),
    "Random Forest (FS)": Pipeline([("prep", preprocessor_fs),
                                     ("clf", RandomForestClassifier(n_estimators=300, max_depth=8, min_samples_leaf=10,
                                                                     class_weight="balanced", random_state=RANDOM_SEED))]),
}

fs_rows = []
for name, pipe in models_fs.items():
    pipe.fit(X_train[selected_original], y_train)
    y_pred = pipe.predict(X_test[selected_original])
    y_proba = pipe.predict_proba(X_test[selected_original])[:, 1]
    fs_rows.append({"Model": name, "Accuracy": accuracy_score(y_test, y_pred), "Precision": precision_score(y_test, y_pred, zero_division=0),
                     "Recall": recall_score(y_test, y_pred), "F1": f1_score(y_test, y_pred), "ROC-AUC": roc_auc_score(y_test, y_proba)})

fs_summary = pd.DataFrame(fs_rows).set_index("Model").round(3)
pd.concat([test_summary, fs_summary])
```

Con un terzo delle feature originali le prestazioni restano nello stesso intervallo osservato con il set completo, senza un miglioramento sistematico. Se il problema fosse rumore introdotto da feature poco informative, rimuoverle avrebbe dovuto migliorare le metriche in modo netto: cosa che non succede.

### 12.2 Confronto con una rete neurale (MLP)

Aumentiamo la capacità del modello con un Multi-Layer Perceptron. `MLPClassifier` non supporta nativamente `class_weight`: bilanciamo il training set con oversampling della classe minoritaria.

```python
# confrontiamo con una rete neurale per avere un confronto con un metro differnte

train_df = X_train.copy()
train_df["defect_label"] = y_train.values
majority = train_df[train_df["defect_label"] == 0]
minority = train_df[train_df["defect_label"] == 1]
minority_upsampled = resample(minority, replace=True, n_samples=len(majority), random_state=RANDOM_SEED)
train_balanced = pd.concat([majority, minority_upsampled]).sample(frac=1, random_state=RANDOM_SEED)

X_train_bal = train_balanced.drop(columns=["defect_label"])
y_train_bal = train_balanced["defect_label"]

mlp_pipe = Pipeline([
    ("prep", preprocessor),
    ("clf", MLPClassifier(hidden_layer_sizes=(32, 16), activation="relu", solver="adam", alpha=0.01,
                           early_stopping=True, validation_fraction=0.15, n_iter_no_change=15,
                           max_iter=500, random_state=RANDOM_SEED))
])
mlp_pipe.fit(X_train_bal, y_train_bal)

y_pred_mlp = mlp_pipe.predict(X_test)
y_proba_mlp = mlp_pipe.predict_proba(X_test)[:, 1]

mlp_summary = pd.DataFrame([{
    "Model": "MLP (32,16)", "Accuracy": accuracy_score(y_test, y_pred_mlp),
    "Precision": precision_score(y_test, y_pred_mlp, zero_division=0), "Recall": recall_score(y_test, y_pred_mlp),
    "F1": f1_score(y_test, y_pred_mlp), "ROC-AUC": roc_auc_score(y_test, y_proba_mlp),
}]).set_index("Model").round(3)

pd.concat([test_summary, mlp_summary])
```

Il MLP non supera il ROC-AUC dei tre modelli originari, nonostante una capacità nominale superiore a Decision Tree e Logistic Regression. Coerente con quanto osservato in sezione 7: passare da un modello lineare a un ensemble di alberi non aveva già prodotto un miglioramento netto; passare a una rete neurale non lo produce nemmeno ora.

### 13.2 Modello consigliato

Non esiste un vincitore assoluto tra i tre modelli di riferimento:
- **Logistic Regression** (L2, C basso) se la priorità è minimizzare i falsi negativi — F1 e recall più alti, coerente con l'alto costo di un richiamo OEM;
- **Random Forest** se si vuole sfruttare la sua migliore capacità di ranking del rischio (ROC-AUC più alto) all'interno del sistema a tre zone di sezione 10, piuttosto che come classificatore binario.

### 13.3 Limiti riscontrati

1. Il potere predittivo attuale non giustifica un'automazione spinta della decisione di scarto (sezione 10): con falsi negativi costosi, la soglia "sicura" tende a richiedere il controllo di quasi tutti i pezzi.
2. Le variabili disponibili descrivono *cosa* è stato misurato, non *quanto* ogni misura si discosti dalla tolleranza ingegneristica specifica — un'informazione che il modello non può dedurre da solo, e che le verifiche di robustezza (sezione 12) confermano non essere recuperabile né con meno feature né con un modello più complesso.
3. `material_batch` e `operator_id`, pur trattati correttamente, non emergono come leve forti; servirebbero più osservazioni per lotto/operatore prima di escluderli operativamente.

### 13.4 Raccomandazioni operative

- Adottare il modello come **sistema di triage a tre zone** (accettazione automatica / controllo 100% / scarto-blocco), calibrando le soglie sui costi reali di FN e FP forniti da AutomaParts, non sulle stime illustrative usate qui.
- Dare priorità di manutenzione/calibrazione alle linee con tasso di difettosità sistematicamente superiore alla media (sezione 2).
- Spostare l'intervento a monte del processo: `cycle_time_s` e `vibration_level` sono le variabili più informative e sono misurabili in tempo reale durante la lavorazione. Un allarme diretto su quando questi parametri escono dal range atteso ha un rapporto costo/beneficio più immediato di una classificazione a fine linea.
- Ricalibrare il modello periodicamente ed obbligatoriamente ad ogni cambio di fornitore di materiale, poiché un segnale già debole è più vulnerabile al data drift.

### 13.5 Sviluppi futuri

- **Limiti di tolleranza ingegneristica per misura**: sapere qual è il range accettabile per ogni misura, per calcolare quanto un pezzo si scosta da quel range invece del solo valore assoluto. È la leva più diretta contro l'high bias di sezione 8.
- **Causa del difetto, non solo presenza/assenza**: codificare che tipo di difetto è (planarità, coppia, temperatura...), non solo difettoso sì/no, per modelli diagnostici e non solo di scarto.
- **Storico di manutenzione/calibrazione** degli strumenti di misura, per linea e per stazione — per distinguere difetti reali da derive strumentali.
- **Serie temporali per pezzo**, non un valore puntuale: l'andamento di vibrazione/temperatura durante l'intero ciclo, per abilitare allarmi in tempo reale invece che a fine linea.
- **Immagini di ispezione visiva** vere e proprie, da affiancare in futuro con un modello di visione artificiale al punteggio numerico visual_inspection_score già usato.
